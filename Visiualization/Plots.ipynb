# Missing Value Graph:

```python
def missing (df):
    plt.figure(figsize=(12,8))
    sns.heatmap(df.isna())
    plt.title(f'Missing values Graph:')
    plt.tight_layout()
    plt.show()
missing(df)
```

# Box Plot:

```python
def box_plot(df):
    num_cols=df.select_dtypes(include='float64')
    for i in num_cols:
        plt.figure(figsize=(6,4))
        sns.boxplot(data=df,x=i)
        plt.title(f'Box Plot of {i}')
        plt.tight_layout()
        plt.show()
box_plot(df)
```

.

.

# Hist Plot:

```python
def hist_plot(df):
    num_cols=df.select_dtypes(include='float64')
    for i in num_cols:
        plt.figure(figsize=(6,4))
        sns.histplot(data=df,x=i,kde=True)
        plt.title(f'Histplot of {i}')
        plt.tight_layout()
        plt.show()
hist_plot(df)
```

.

.

# Corelation Matrix:

```python
def corelation(df):
    plt.figure(figsize=(8,6))
    sns.heatmap(df.corr(numeric_only=True),annot=True,cmap='coolwarm')
    plt.title(f'Corelation Matrix:')
    plt.tight_layout()
    plt.show()
corelation(df)    
```

# Pair Plot:

```python
def pair_plot(df):
    plt.figure(figsize=(12,10))
    sns.pairplot(df, diag_kind='kde')
    plt.title(f'Pair Plot:')
    plt.tight_layout()
    plt.show()
pair_plot(df)    
```

# Line Plot:

```python
def line_plot(df):
    plt.figure(figsize=(8,6))
    sns.lineplot(data=df,x='ConvertedCompYearly',y='JobSat')
    plt.title(f'Line Plot of Salary vs Job Satisfaction:')
    plt.tight_layout()
    plt.show()
line_plot(df)    
```

# Scatter Plot:

```python
def scatter_plot(df):
    plt.figure(figsize=(8,6))
    sns.scatterplot(data=df,x='ConvertedCompYearly',y='JobSat')
    plt.title(f'Scatter Plot of Salary vs Job Satisfaction:')
    plt.tight_layout()
    plt.show()
scatter_plot(df)    
```
