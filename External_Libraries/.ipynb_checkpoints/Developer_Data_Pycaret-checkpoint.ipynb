```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/Developer_data')
df.sample()
```

```python
df2=df[df['ConvertedCompYearly']<250000]
```

```python
pip install pycaret
```

```python
from pycaret.regression import setup, compare_models

regression_setup = setup(data=df2, target='ConvertedCompYearly', session_id=1122)
```

```python
best_model = compare_models()
```

```python
from pycaret.regression import tune_model

tuned_best_model = tune_model(best_model)
print(tuned_best_model.get_params())
```
